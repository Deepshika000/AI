```python
#Install Hugging Face Transformers
#It's simply a Hugging Face library that helps us load datasets
!pip install transformers datasets

from transformers import pipeline
#Creating sentiment tool
sentiment_pipeline = pipeline("sentiment-analysis")
pipeline("sentiment-analysis") #print this
```

```python
#Mysentences
text = "I absolutely love this Recipe!"
result = sentiment_pipeline(text)

print(result)
#here model will predict sentiment based on the sentence ,in this sentence there is no negative word
```

```python
#Creating same dataset

import pandas as pd

data = {"text": [
        "I love this product",
        "This product is amazing",
        "I really enjoyed this movie",
        "The service was excellent",
        "I hate this product",
        "This product is terrible",
        "I did not enjoy this movie",
        "The service was horrible"],

    "label": [
        "POSITIVE",
        "POSITIVE",
        "POSITIVE",
        "POSITIVE",
        "NEGATIVE",
        "NEGATIVE",
        "NEGATIVE",
        "NEGATIVE"]}

df = pd.DataFrame(data)

print(df)
```

```python
# Lets Running Hugging Face on every text

predictions = sentiment_pipeline(df["text"].tolist()) #simply converts the pandas column into a normal Python list

print(predictions) # printthose list
```

```python
#Evaluating the model
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(df["label"],df["predicted"])
print("Accuracy:", accuracy)
```

```python
#Lets convert Precision, Recall, F1 into dataframe

from sklearn.metrics import classification_report

print(classification_report(df["label"],df["predicted"]))
```
